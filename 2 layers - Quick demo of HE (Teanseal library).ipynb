# Some Parameters
$R_t = \mathbb{Z}_t[x]/(x^n + 1) ⇒$ Plaintext polynomial. \\
$R_q = \mathbb{Z}_q[x]/(x^n + 1) ⇒$ Ciphertext polynomial. \\
$n ⇒$ Polynomial modulus degree. \\
Find primes of length specifiend in the array "coeff_mod_bit_sizes". For instance, $p_0, p_1, p_2, \cdots, p_k$ \\
Then, $q = q_k = p_1 \times p_2 \times \cdots \times p_k$ \\
Similarly, $t = p_0$ \\

$Δ = \frac{q_k}{q_{k-1}} \Rightarrow$ Scale


```python
# parameters
# poly_mod_degree = 32768
# coeff_mod_bit_sizes = [40, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 40]

# parameters
# poly_mod_degree = 16384
# coeff_mod_bit_sizes = [40, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 25, 40]

# parameters
# poly_mod_degree = 8192
# coeff_mod_bit_sizes = [30, 21, 21, 21, 30]

# parameters
poly_mod_degree = 4096
coeff_mod_bit_sizes = [30, 21, 21, 30]

# create TenSEALContext
context = ts.context(ts.SCHEME_TYPE.CKKS, poly_mod_degree, -1, coeff_mod_bit_sizes)
context.global_scale = 2 ** 21
context.generate_galois_keys()
```

```python
x = np.array([1.1, 2.2, 3.3])
y = np.array([6.6, 5.5, 4.4])

x_enc = ts.ckks_vector(context, x)
y_enc = ts.ckks_vector(context, y)

result_enc = x_enc + y_enc

result_decrypted = result_enc.decrypt()

print(f'Original result: {x + y}')
print(f'Decrypted result: {result_decrypted}')
print(f'Difference: {result_decrypted - x - y}')
```

```python
# parameters
poly_mod_degree = 8192
coeff_mod_bit_sizes = [30, 21, 21, 21, 30]

# create TenSEALContext
context = ts.context(ts.SCHEME_TYPE.CKKS, poly_mod_degree, -1, coeff_mod_bit_sizes)
context.global_scale = 2 ** 21
context.generate_galois_keys()
```

```python
x = np.array([1.1, 2.2, 3.3])
x_enc = ts.ckks_vector(context, x)

result_enc = x_enc * x_enc * x_enc * x_enc
print(f'Original reult: {x * x * x}')
print(f'Decrypted result: {result_enc.decrypt()}')
```
